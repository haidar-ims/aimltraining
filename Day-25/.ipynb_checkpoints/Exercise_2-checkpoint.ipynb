### Bag of Words

```python
import pandas as pd
import numpy as np
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
import re
import nltk
from nltk.corpus import stopwords
from sklearn.linear_model import LogisticRegression
```

```python
nltk.download('stopwords')
nltk.download('wordnet')
```

```python
data = {
    "review": [
        "I absolutely love this phone! The camera quality is fantastic and battery lasts all day.",
        "Terrible experience. The screen broke within a week, and customer service was useless.",
        "Great value for money. Performance is smooth and build quality feels premium.",
        "The laptop overheats and the battery drains too fast. Very disappointed.",
        "Excellent sound quality and very comfortable to wear. Highly recommend these headphones!",
        "Worst product I’ve ever bought. It stopped working after two days.",
        "Amazing performance and display. Totally worth the price.",
        "Poor design and fragile build. I regret buying this item.",
        "Fast delivery and great packaging. The product works as expected.",
        "Customer support was unhelpful. Will not buy from this brand again."
    ],
    "sentiment": [
        "positive", "negative", "positive", "negative", "positive",
        "negative", "positive", "negative", "positive", "negative"
    ]
}
```

```python
 df = pd.DataFrame(data)
 df.head()
```

```python
# Function to clean and lower data
def clean_text(text):
    text = re.sub(r'[^\w\s]', '', text)
    text = text.lower()
    text = ' '.join([word for word in text.split() if word not in stopwords.words('english')])
    return text
```

```python
df['clean_review']=df['review'].apply(clean_text)
df
```

```python
cv = CountVectorizer(max_features=20)
X = cv.fit_transform(df['clean_review']).toarray()
y = df['sentiment']

print("Top 25 words in the feature X")
cv.get_feature_names_out()

bow_df=pd.DataFrame(X,columns=cv.get_feature_names_out())
bow_df
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
model = LogisticRegression()
model.fit(X_train, y_train)
```

```python
# --- User Prediction ---

# Taking review from user
user_review = input("Enter your product review: ")

# Clean the review
# Assuming you have a function `clean_text()` defined earlier
user_review = clean_text(user_review)

# Convert cleaned review to BOW representation
user_review_bow = cv.transform([user_review]).toarray()

# Predict using the trained model
user_sentiment = model.predict(user_review_bow)

# Output predicted sentiment
print("\nCleaned Review:", user_review)
print("Predicted Sentiment:", user_sentiment[0])
```

Do the following Tasks:
1. Spilts text into words/tokens (show the individual word token)
2. Clean Tokens (show clean token)
3. Remove Stop words (Show token after removing stop words)
4. Stemming (Show the words in their root forms)
5. Lemmatization (Show lemma words)

```python
import nltk
# import spacy
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import PorterStemmer
from nltk.stem import WordNetLemmatizer
import re
# from sklearn.feature_extraction.text import CountVectorizer
from sklearn.feature_extraction.text import TfidfVectorizer
import pandas as pd
```

```python
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')
nltk.download('punkt_tab')
```

```python
input_text = "Natural Language Processing is going to help you out in near future . You have 2 remember this!! for getting interview calls"
```

```python
# 1. Spilts text into words/tokens (show the individual word token)

tokens = word_tokenize(input_text) # Split text into individual
print(tokens)
```

```python
# 2. Clean Tokens (show clean token)

clean_tokens=[re.sub('[^a-zA-Z]','', token. lower()) for token in tokens if token.isalpha()]
print(clean_tokens)
```

```python
# 3. Remove Stop words (Show token after removing stop words)

#Stop Word Removal
stop_words=set(stopwords.words('english'))
filtered_tokens=[token for token in clean_tokens if token not in stop_words]
print(filtered_tokens)
```

```python
# 4. Stemming (Show the words in their root forms)

# Stemming
stemmer=PorterStemmer()
stemmed_tokens=[stemmer.stem(token) for token in filtered_tokens]
print(stemmed_tokens)
```

```python
# 5. Lemmatization (Show lemma words)

lemmatizer = WordNetLemmatizer()
lemmatized_tokens = [lemmatizer.lemmatize(token) for token in filtered_tokens]
# Convert words to their dictionary form (lemma) using POS: Part of Speech
print(lemmatized_tokens)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
import pandas as pd

# --- Step 1: Initialize TF-IDF Vectorizer ---
# TF-IDF = Term Frequency – Inverse Document Frequency
# It converts text data into numeric values based on how important each word is across documents.
tfidf = TfidfVectorizer()

# --- Step 2: Fit and transform your text data ---
# Here, 'lemmatized_tokens' should be a list of complete text documents (strings),
# not just individual words.
matrix = tfidf.fit_transform(lemmatized_tokens)

# --- Step 3: View the TF-IDF matrix ---
# 'matrix' is a sparse matrix (efficient for large text data).
matrix

# --- Step 4: Convert to DataFrame for readability ---
# Converts the sparse matrix into a table format (rows = documents, columns = words)
df = pd.DataFrame(matrix.toarray(), columns=tfidf.get_feature_names_out())

# --- Step 5: Display the DataFrame ---
df
```

## TF-IDF Example-2

```python
corpus=[
    "Python developer required with strong coding skills ",
    " Experience in REST API, Automation, Jenkins, Flask and Python",
    "Looking for machine learning engineer familiar with NLP, Supervised, Unsupervised and Deep Learning "
]
```

```python
tfidf = TfidfVectorizer()

matrix = tfidf.fit_transform(corpus)
df = pd.DataFrame(matrix.toarray(), columns=tfidf.get_feature_names_out())
print(df)
```

```python
print('\n Top Weighted Terms')
print(df.mean().sort_values(ascending=False).head(10))
```

#### Difference Between CountVectorizer and TfidfVectorizer
| Feature                    | **CountVectorizer**                                       | **TfidfVectorizer**                                                             |
| -------------------------- | --------------------------------------------------------- | ------------------------------------------------------------------------------- |
| **Full Name**              | Bag of Words (Count Vectorizer)                           | Term Frequency–Inverse Document Frequency (TF-IDF)                              |
| **What it does**           | Counts how many times each word appears in each document. | Measures how *important* each word is in a document, compared to all documents. |
| **Output Values**          | Simple integer counts (e.g. 0, 1, 2, 3...)                | Decimal weights (e.g. 0.23, 0.89...) representing importance.                   |
| **Effect of Common Words** | Common words like “the”, “is”, “and” may dominate.        | Common words get *lower weight* because they appear in many documents.          |
| **Formula (conceptually)** | `Count(word)`                                             | `TF × IDF = (Term Frequency) × (Inverse Document Frequency)`                    |
| **When to Use**            | Small datasets or when raw frequency matters.             | When you need to focus on *informative* or *unique* words.                      |
| **Example (word “data”)**  | Appears 5 times → count = 5                               | Appears 5 times but in every document → low TF-IDF (less important)             |
