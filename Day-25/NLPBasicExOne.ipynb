import re

tokens = ["Apple", "is", "looking", "at", "buying", "a", "startup", "in", "the", "UK", "!", "can't", "email123"]

clean_tokens = []

for token in tokens:
    # Remove all non-letter characters but keep letters and apostrophes
    cleaned = re.sub(r"[^a-zA-Z']", "", token)
    
    # Convert to lowercase and ignore empty strings
    if cleaned:
        clean_tokens.append(cleaned.lower())

print(clean_tokens)


```python
#Stop Word Removal
stop_words=set(stopwords.words('english'))
filtered_tokens=[token for token in clean_tokens if token not in stop_words]
filtered_tokens
```

```python
# Stemming
stemmer=PorterStemmer()
stemmed_tokens=[stemmer.stem(token) for token in filtered_tokens]
stemmed_tokens
```

```python
lemmatizer = WordNetLemmatizer()
lemmatized_tokens = [lemmatizer.lemmatize(token) for token in filtered_tokens]
# Convert words to their dictionary form (lemma) using POS: Part of Speech
lemmatized_tokens
```

```python
corpus = [
    "Apple is looking at buying a startup in the UK.",
    "Google and Microsoft are competitors in the tech industry.",
    "I love AI and machine learning."
]
```

```python
# Create CountVectorizer
cv = CountVectorizer() # This class converts a collection of text documents (corpus) into a matrix of token counts—the Bag-of-Words representation.
bow_matrix = cv.fit_transform(corpus)

# Print feature names
print("********* Feature Names (BOW) *********")
print(cv.get_feature_names_out())

# Print BOW matrix as array
print("********* BOW Matrix Array *********")
print(bow_matrix.toarray())
```
