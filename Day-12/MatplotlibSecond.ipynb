# We have following dictionary
    ## product_sales={'Laptop': 750, 'Smart Phone: 650, 'Smart Watch': 400, 'Tablets': 500, 'AC':400}
    
### Repersent this data using bar chart xlabel is ProductName (e.g. Laptop) and ylable is sales value (e.g. 750)

```python
product_sales = {
    'Laptop': 750,
    'Smart Phone': 650,
    'Smart Watch': 400,
    'Tablet': 500,
    'AC': 400
}

products=['Laptop','Smartphones','Smart Watch','Tablet','AC']
weekly_sales=[750,650,400,500,400]

plt.figure(figsize=(8, 5))
bar_colors=['blue','red','green','yellow','orange','purple','black']

plt.bar(products,weekly_sales,color=bar_colors,width=0.5,edgecolor='black')

plt.title('Weekly Products Sales')
plt.xlabel('Products')
plt.ylabel('Sales')
plt.tight_layout()
plt.show()
```

### Use data directly from your dictionary (less typing):

```python
plt.bar(product_sales.keys(), product_sales.values(), color=bar_colors)
plt.figure(figsize=(12, 7))
```

# Scatter Plots
### In scatter plots, we represent data in points, where each point represent a data observation, we can check how to values correlate

```python
product_sales={'Laptop":750, "Smart Phone': 650, 'Smart Watch':400, 'Tablets': 500, 'AC':400}
products=list(product_sales.keys())
sales=list(product_sales.values())
plt.scatter(products, sales)
plt.title('Basic Product and Sales Representation')
plt.xlabel('Products')
plt.ylabel('Sales')
plt.grid('True')
plt.show()
```

```python
players=['MSD', 'Rohit', 'Virat', 'Gill', 'Shami']
match1_score=[45,67,56,78,89]
match2_score=[55,78,40,80,97]
match3_score=[99,40,20,30,60]
plt.figure(figsize=(8,5))
plt.scatter(players,match1_score,color='red', label='Match One', s=100)
plt.scatter(players,match2_score,color='green', label='Match Two', s=100, marker='x')
plt.scatter(players,match3_score,color='blue', label='Match Three', s=100, marker='p')
plt.title('Comparison of Players in Two Matches')
plt.xlabel('Players')
plt.ylabel('Scores')
plt.legend(loc='best')    # multiple loc like centre. upper right and so on
plt.grid(True,linestyle='--', alpha=0.5)
plt.show()
```

### Write a comparison that shows average calories burned for each workout type from final_data dataset

```python
df=pd.read_csv('Final_data.csv')
avg_cal=df.groupby ('Workout_Type') ['Calories_Burned'].mean()
selected_index=avg_cal.index
selected_value=avg_cal.values
plt.scatter(selected_index, selected_value, color='green', s=100, marker='p')
plt.title('Comparision of Workout and calories Burned')
plt.xlabel('Workout Type')
plt.ylabel('Calories Burned')
plt.grid(True, linestyle='-', alpha=0.5)
plt.show()
```
