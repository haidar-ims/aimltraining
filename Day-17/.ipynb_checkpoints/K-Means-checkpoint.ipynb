#### K-Means Clustering
    Example

    K-means comes under unsupervised Machine Learning Technique

    Terms we will use here, like Cluster

    Centroid

    Cluster: we will partition our dataset into K distinct clusters

    centroid: nearest mean cluster

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.cluster import KMeans
```

```python
X = np.array([
    [7.5, 8.0],
    [9.0, 7.0],
    [10.5, 6.5],
    [6.0, 9.0],
    [13.0, 5.5],
    [8-0, 8.5],
    [14.0, 4.5],
    [9.5, 7.0],
    [7.0, 9.0],
    [12.0, 6.0]
])

df = pd.DataFrame(X, columns=['EntryTime' ,'Duration' ])
df
```

```python
#create model
kmeans = KMeans(n_clusters=3, random_state=0)
```

```python
df['Cluster'] = kmeans.fit_predict(df[['EntryTime','Duration']])
centers = kmeans. cluster_centers_
df
```

```python
plt.figure(figsize=(6,5))
plt.scatter(df['EntryTime'], df['Duration'], c=df['Cluster' ])
plt.scatter(centers[:,0], centers[: , 1], marker='X', s=200)
plt.xlabel('Entry Time (Hour of Day)')
plt.ylabel('Duration in Office (Hours)')
plt.title('Working Hour Patterns using K-Means')
plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(7,5))

# Scatter plot of data points colored by cluster
plt.scatter(df['EntryTime'], df['Duration'], c=df['Cluster'], cmap='viridis', s=100)

# Scatter plot of cluster centroids
plt.scatter(centers[:,0], centers[:,1], marker='X', s=200, c='red', label='Centroid')

# Labels and title
plt.xlabel('Entry Time (Hour of Day)')
plt.ylabel('Duration in Office (Hours)')
plt.title('Working Hour Patterns using K-Means')

# Legend and grid
plt.legend()
plt.grid(True)

# Show plot
plt.show()
```

### Exercise
#### Lab: Customer Segmentation using K-Means
    Goal: Group customers based on Annual Income and Spending Score.
    
    Steps:

    1. Download customers.csv from GitHub.
    2. Load the data from customers.csv.
    3. Visualize the data.
    4. Apply K-Means clustering.
    5. Interpret each segment, e.g., “High income / High spender,” etc.
    6. Create a scatter plot to visualize the clusters.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.cluster import KMeans
```

```python
# Load dataset
customers = pd.read_csv('customers.csv')
print(customers.head())
```

```python
# Select features for clustering
X = customers[['AnnualIncome', 'SpendingScore']]
```

```python
# Create and fit K-Means
kmeans = KMeans(n_clusters=5, random_state=0)
labels = kmeans.fit_predict(X)

# Add cluster labels to DataFrame
customers['Cluster'] = labels

# Get cluster centers
centers = kmeans.cluster_centers_

# Show first few rows
print(customers.head())
```

```python
# Plot clusters with centroids
plt.figure(figsize=(8,6))

# Scatter plot of customers colored by cluster
plt.scatter(customers['AnnualIncome'], customers['SpendingScore'], 
            c=customers['Cluster'], cmap='viridis', s=100)

# Scatter plot of cluster centroids
plt.scatter(centers[:,0], centers[:,1], marker='X', s=200, c='red', label='Centroid')

# Labels and title
plt.xlabel('Annual Income (MYR)')
plt.ylabel('Spending Score (1-100)')
plt.title('Customer Segments (K-Means)')

# Legend and grid
plt.legend()
plt.grid(True)

# Show plot
plt.show()
```
