#### Exercise: Smart Traffic Light Controller using Q-Learning
    Objective: In this exercise, you'll design a Smart Traffic Light System that learns when to switch lights (Green/Red) based on real-time traffic conditions using Q-Learning. Your Al agent will balance reducing waiting time for vehicles, saving energy, and maintaining safety

```python
# Step 1: Import Libraries
import numpy as np
import random
```

```python
# Step 2: Define States and Actions

# Traffic states
# 0=Empty, 1=Light, 2=Moderate, 3=Heavy, 4=Very Heavy
states = [0, 1, 2, 3, 4]

# Actions
actions = ['GREEN', 'RED']

print("States:", states)
print("Actions:", actions)
```

```python
# Step 3: Initialize Q-Table and Hyperparameters
Q = np.zeros((len(states), len(actions)))
# Q-table stores "quality" (Q) of each action at each state, all start at 0

alpha = 0.1      # Learning rate
gamma = 0.9      # Discount factor
epsilon = 0.2    # Exploration rate
episodes = 300   # Training runs

print('Q-table shape:', Q.shape)
```

```python
# Step 4: Reward Functio
def get_reward(states, action):
    if states >= 3 and action == 'GREEN':
        return 5
    elif states <= 2 and action == 'RED':
        return 5
    else:
        return -10
print ('Reward Example (0=low, GREEN):',get_reward(4, 'RED'))
```

```python
# Step 5: Environment Dynamics
def next_traffic(current):
    new = current + random.choice([-1, 0, 1])
    return int(np.clip(new, 0, 4))
```

```python
# Step 6: Train the Agent
for ep in range(episodes):
    traffic = random.choice(states) #Tweak

    for _ in range(15):
        if random.uniform(0, 1) < epsilon:   # Epsilon-greedy startegy
            action = random.choice(actions)  
        else:
            action = actions[np.argmax(Q[traffic])]  #Tweak

        next_state = next_traffic(traffic)
        reward = get_reward(next_state, action)

        # Q-learning update
        a = actions.index(action)
        best_next = np.max(Q[next_state])              # Tweak
        Q[traffic, a] += alpha * (reward + gamma * best_next - Q[traffic, a])   #Tweak

        traffic = next_state

print("Training complete!")
```

```python
# Step 7: Test the Learned Traffic Light Controller
try:
    traffic = int(input("Enter starting traffic (0-4): "))
    if traffic < 0 or traffic > 4:
        raise ValueError("Traffic must be 0-4")
except ValueError as e:
    print(e)
    traffic = 2
    print("Default traffic set to 2 (Moderate)")

print("\nSimulating traffic light for 10 steps:\n")
for step in range(10):
    action = actions[np.argmax(Q[traffic])]
    print(f"Step {step+1}: Traffic={traffic} -> Action={action}")
    traffic = next_traffic(traffic)

print("\nSimulation finished!")
```
