# .loc: Label Based
# .iloc: Index Based

```python
print('original dataframe')
print(df)
print('Location : Label Based')
print(df.loc[2,['Name','Marks']])
```

```python
print('original dataframe')
print(df)
print('Location : Index Based')
print(df.iloc[2,[1,2]])
```

# Axis in Numpy : 
    1. Axis 0 : vertical direction works column-wise   
    2. Axis 1 : horizontal direction: works row-wise

# We have two students' marks in three subjects
[[45,50,49],[30,45,49]]

```python
marks_arr=np.array([[45,50,49],[30,45,49]])
print(marks_arr)
```

##### 📘 Axis 0 → column-wise (top to bottom)

```python
print('Sum down each column',np.sum(marks_arr,axis=0))
```

##### 📘 Axis 1 → row-wise (left to right)

```python
print('Sum down each row',np.sum(marks_arr,axis=1))
```

# Find the student name who got highest mark

```python
topper=df.loc[df['Marks'].idxmax(),'Name']
print('Topper Student Name',topper)
```

```python
print('Next Example')
```

```python
top_index = df['Marks'].idxmax()
top_student=df.iloc[top_index,df.columns.get_loc('Name')]
print('Topper of the class is: ',top_student)
```

| Step                         | Purpose                         | Example Result                 |
| ---------------------------- | ------------------------------- | ------------------------------ |
| `df['Marks'].idxmax()`       | Find index of highest marks     | `1`                            |
| `df.columns.get_loc('Name')` | Find position of “Name” column  | `0`                            |
| `df.iloc[...]`               | Select that cell (row 1, col 0) | `"Bala"`                       |
| `print(...)`                 | Show the result                 | “Topper of the class is: Bala” |


##### 💬 Plain English version:

“Look at the Marks column, find the row with the highest mark,
then go to that row and read the student’s Name.”

    # df.iloc[row_index, column_index] → takes row first, then column.

    # op_index is the row index (for the student with highest marks).

    # df.columns.get_loc('Name') gives the column index number for 'Name'.
    So together, this means “find the value at (row = top student, column = Name).”

💡 Summary:

| Method  | Type                  | Syntax Example               | Uses                       |
| ------- | --------------------- | ---------------------------- | -------------------------- |
| `.iloc` | index-based (numbers) | `df.iloc[2, 1]`              | Row 2, Col 1               |
| `.loc`  | label-based (names)   | `df.loc['RowLabel', 'Name']` | Row “RowLabel”, Col “Name” |
