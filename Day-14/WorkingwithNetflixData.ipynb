## Make a chart with Seaborn’s histplot() using the Netflix data. Show the release years of Movies and TV Shows in different colors.

```python
plt.figure(figsize=(10,5))
sns.histplot(
    data=df,
    x='release_year',
    hue='type',
    bins=30,
    multiple='dodge',
    palette={'Movie': 'green', 'TV Show': 'red'},
    edgecolor='yellow',
    alpha=0.8
)

plt.title('Netflix Movies vs TV Shows Released by Year')
plt.xlabel('Release Year')
plt.ylabel('Number of Titles')

plt.tight_layout()
plt.savefig('netflix_movies_vs_tvshows_by_year.png',dpi=300, bbox_inches='tight')
plt.show()
```
