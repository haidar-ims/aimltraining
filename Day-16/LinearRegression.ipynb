# Supervised Machine Learning
## Linear Regression

```python
import pandas as pd
from sklearn.linear_model import LinearRegression
import numpy as np
```

```python
data = pd.read_csv("sales_data.csv")
data.head()
```

```python
X = data[['Advertising_Spend']]
y = data['Sales_Revenue']
```

```python
model = LinearRegression()   # create the model
model.fit(X, y)              # train the model
```

```python
new_spend = int(input('Enter New Spend to Predict Sales Revenue'))
predicted_revenue = model.predict([[new_spend]])
print(f'New Spend {new_spend} Predicted Revenue {predicted_revenue}')
```

    # Goal: Train a Linear Regression model to predict house price from size.

```python
import pandas as pd
from sklearn.linear_model import LinearRegression
import numpy as np

# Load dataset
df = pd.read_csv('house_price.csv')
df
```

```python
# Replace "?" with actual NaN
df.replace('?', np.nan, inplace=True)
```

```python
df=df.dropna()
print('Dataset after dropna()')
df
```

```python
X = df[['Size_sqft']]
y = df['Price']
```

```python
model = LinearRegression()
model.fit(X,y)
```

```python
# Step 7: Make predictions
pr_price = model.predict([[1700]])
print(pr_price)
```
