# KNN Algorithm 
#### KNN Algorithm in Supervised Machine Learning
#### KNeighbors Classifier of sklearn libray for this one

```python
import numpy as np
from sklearn.neighbors import KNeighborsClassifier
```

```python

X = np.array([
    [1, 3500],  # Unfit
    [2, 3200],  # Unfit
    [3, 3000],  # Unfit
    [4, 2500],  # Fit
    [5, 2300],  # Fit
    [6, 2000],  # Fit
    [7, 2200],  # Fit
    [2, 3700]   # Unfit
])

y = ['Unfit', 'Unfit', 'Unfit', 'Fit', 'Fit', 'Fit', 'Fit', 'Unfit']
```

```python
# Create model
model = KNeighborsClassifier(n_neighbors=3)
```

```python
# Train model
model.fit(X,y)
```

```python
#predict
new_person = [[4.5, 2600]] # 4.5 hrs exercise, 2600 calories
pred = model.predict(new_person)
print("Predicted Result:", pred[0])
```

```python
# Prediction from taking the values from users
exercise_hours=float(input('Enter number of Exercise Hours'))
cal_intake=float(input('Enter number of calories consumed'))
pred_fu=model.predict([[exercise_hours,cal_intake]])
print(f'Exercise Hours {exercise_hours}, Calories Intake {cal_intake} You are likely {pred_fu[0]}')
```

```python
import pandas as pd
```

```python
df = pd.read_csv("fruit_data.csv")
df
```

```python
X = df[['Weight','Size']]
y = df['Fruit']
```

```python
model = KNeighborsClassifier(n_neighbors=3)
```

```python
model.fit(X,y)
```

#### Closing 1

```python
# new_fruit = [[120, 2]]  # Example: weight = 120 gm, size = 2 cm
new_fruit = pd.DataFrame([[120, 2]], columns=['Weight', 'Size'])
pred = model.predict(new_fruit)
print("Predicted Result:", pred[0])
```

#### Closing 2

```python
fruit_weight = float(input('Enter weight of fruit in gm: '))
fruit_size = float(input('Enter size of fruit in cm: '))

# Create a DataFrame with correct column name syntax
our_fruit = pd.DataFrame([[fruit_weight, fruit_size]], columns=['Weight', 'Size'])

# Predict using your trained model
pred_fruit = model.predict(our_fruit)

# Print the result using proper f-string
print(f'Fruit Weight is {fruit_weight}, Fruit size is: {fruit_size}. The Fruit is likely: {pred_fruit[0]}')
```

### Exercise
#### Download student_grade.csv from GitHub.com.
#### Use the K-Nearest Neighbors (KNN) method to predict a student’s grade (A, B, or C) based on:
    # Hours studied
    # Assignments completed

```python
import pandas as pd
import numpy as np
from sklearn.neighbors import KNeighborsClassifier
```

```python
df = pd.read_csv("student_grade.csv")
df
```

```python
X = df[['Hours_Studied', 'Assignments_Completed']]

model = KNeighborsClassifier(n_neighbors=3)
model.fit(X, df['Grade'])
```

```python
new_hours = float(input("Enter the Hour of Study"))
new_asgn = float(input("How Much Assignment Completed?"))
pred_grade = model. predict(pd.DataFrame([[new_hours, new_asgn]], columns=['Hours_Studied','Assignments_Completed']))
print(f'Study Hour Time is {new_hours } and Assignment Completed is {new_asgn}. The Grade is likely {pred_grade[0]}')
```
