# What is Matplotlib?
### Matplotlib is a Python library used to create charts and graphs. It’s very useful for data visualization

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]      # X-axis values
y = [10, 20, 25, 30, 50] # Y-axis values

plt.plot(x, y)            # draw line
plt.show()                # display chart
```

### Add Titles and Labels

```python
plt.plot(x, y)
plt.title('Sales Over Time')
plt.xlabel('Months')
plt.ylabel('Sales ($)')
plt.show()
```

### Add Colors, Line Styles, and Markers

```python
plt.plot(x, y, color='red', linestyle='--', marker='o')
```

## Bar Chart

```python
items = ['Laptop', 'Phone', 'Watch']
sales = [750, 650, 400]

plt.bar(items, sales, color='skyblue')
plt.title('Product Sales')
plt.xlabel('Products')
plt.ylabel('Sales')
plt.show()
```

### Scatter Plot

```python
plt.scatter(items, sales, color='green', s=100)  # s = size
plt.title('Product Sales (Scatter)')
plt.show()
```

### Pie Chart

```python
plt.pie(sales, labels=items, autopct='%1.1f%%')
plt.title('Sales Share')
plt.show()
```

### Multiple Plots on Same Graph

```python
x = [1, 2, 3, 4]
sales_A = [10, 20, 15, 25]
sales_B = [12, 18, 20, 22]

plt.plot(x, sales_A, label='Store A')
plt.plot(x, sales_B, label='Store B')
plt.legend()   # shows label box
plt.show()
```

### Grid, Size, and Style

```python
plt.figure(figsize=(8,5))   # make it bigger
plt.plot(x, sales_A, color='purple')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

## SUBPLOTS (Multiple charts in one figure)

### Two plots side by side

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
sales_A = [10, 20, 15, 25, 30]
sales_B = [12, 18, 20, 22, 28]

plt.figure(figsize=(10,4))   # width=10, height=4

# 1st subplot (rows=1, columns=2, position=1)
plt.subplot(1, 2, 1)
plt.plot(x, sales_A, color='red', marker='o')
plt.title('Store A')

# 2nd subplot (rows=1, columns=2, position=2)
plt.subplot(1, 2, 2)
plt.plot(x, sales_B, color='blue', marker='s')
plt.title('Store B')

plt.tight_layout()  # auto-fix spacing
plt.show()
```

# Matplotlib + Pandas Integration

```python
import pandas as pd
import matplotlib.pyplot as plt

# Create a small dataset
data = {
    'Month': ['Jan', 'Feb', 'Mar', 'Apr', 'May'],
    'Sales': [200, 250, 300, 400, 350],
    'Profit': [50, 80, 100, 130, 120]
}

df = pd.DataFrame(data)
print(df)
```

### Plot Directly from DataFrame Columns

```python
plt.plot(df['Month'], df['Sales'], marker='o', color='blue', label='Sales')
plt.plot(df['Month'], df['Profit'], marker='s', color='green', label='Profit')

plt.title('Sales and Profit Over Months')
plt.xlabel('Month')
plt.ylabel('Amount ($)')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

### Bar Chart from DataFrame

```python
plt.bar(df['Month'], df['Sales'], color='orange', edgecolor='black')
plt.title('Monthly Sales')
plt.xlabel('Month')
plt.ylabel('Sales ($)')
plt.show()
```

### Multiple Bars (Comparison Bars)

```python
x = range(len(df))
plt.bar(x, df['Sales'], width=0.4, label='Sales', color='blue')
plt.bar([i+0.4 for i in x], df['Profit'], width=0.4, label='Profit', color='green')

plt.xticks([i+0.2 for i in x], df['Month'])
plt.title('Sales vs Profit Comparison')
plt.xlabel('Month')
plt.ylabel('Amount ($)')
plt.legend()
plt.show()
```

### Using Pandas Built-in Plotting (Shortcut)

```python
df.plot(x='Month', y=['Sales', 'Profit'], kind='line', marker='o', figsize=(8,5))
plt.title('Sales vs Profit (Using Pandas)')
plt.show()
```

### You can also use:

```python
df.plot(kind='bar', x='Month', y='Sales')
df.plot(kind='pie', y='Sales')
df.plot(kind='scatter', x='Sales', y='Profit')
```

### Realistic Example (Average Calories Burned)

```python
data = {
    'Workout_Type': ['Running', 'Cycling', 'Yoga', 'Swimming'],
    'Avg_Calories': [400, 350, 180, 450]
}

final_data = pd.DataFrame(data)

plt.bar(final_data['Workout_Type'], final_data['Avg_Calories'], color='lightgreen', edgecolor='black')
plt.title('Average Calories Burned by Workout Type')
plt.xlabel('Workout Type')
plt.ylabel('Average Calories Burned')
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.show()
```

### 🌟 Summary

| Goal                | Method                             |
| ------------------- | ---------------------------------- |
| Plot from list      | `plt.plot(x, y)`                   |
| Plot from DataFrame | `plt.plot(df['col1'], df['col2'])` |
| Multiple columns    | `df.plot(x='col', y=['a','b'])`    |
| Bar, Pie, Scatter   | `df.plot(kind='bar')`, etc.        |

# Real-world Data Visualization

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# Create fake company data
data = {
    'Month': ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun'],
    'Sales': [200, 250, 300, 350, 400, 450],
    'Ad_Spend': [20, 25, 40, 30, 50, 55]
}

df = pd.DataFrame(data)
print(df)
```

### Line Chart + Trend Line

```python
plt.figure(figsize=(8,5))
plt.plot(df['Month'], df['Sales'], marker='o', color='blue', label='Sales')

# Add a trend line using NumPy
x = np.arange(len(df))
z = np.polyfit(x, df['Sales'], 1)   # 1 = linear
trend = np.poly1d(z)
plt.plot(df['Month'], trend(x), color='red', linestyle='--', label='Trend Line')

plt.title('Sales Growth Over Months')
plt.xlabel('Month')
plt.ylabel('Sales ($)')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

### Bar + Line Combo Chart

```python
fig, ax1 = plt.subplots(figsize=(8,5))

# Left axis (bar)
ax1.bar(df['Month'], df['Sales'], color='skyblue', label='Sales')
ax1.set_xlabel('Month')
ax1.set_ylabel('Sales ($)', color='blue')

# Right axis (line)
ax2 = ax1.twinx()
ax2.plot(df['Month'], df['Ad_Spend'], color='green', marker='o', label='Ad Spend')
ax2.set_ylabel('Advertising Spend ($)', color='green')

plt.title('Sales vs Ad Spend')
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.show()
```

### ✅ This shows how ad spending might affect sales.

### Histogram (Data Distribution)

```python
data = np.random.normal(70, 10, 100)  # 100 values, mean=70, std=10
plt.figure(figsize=(7,4))
plt.hist(data, bins=10, color='orange', edgecolor='black')
plt.title('Distribution of Test Scores')
plt.xlabel('Score')
plt.ylabel('Number of Students')
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.show()
```

### Scatter Plot (Relationship Between Two Variables)
    Scatter plots help you find correlation between variables:

```python
plt.figure(figsize=(7,5))
plt.scatter(df['Ad_Spend'], df['Sales'], color='purple', s=100, alpha=0.7)
plt.title('Ad Spend vs Sales')
plt.xlabel('Advertising Spend ($)')
plt.ylabel('Sales ($)')
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```

### ✅ If the dots go upward together → sales rise with ad spending → positive correlation.

# Advanced Matplotlib — Professional Charts and Customization

### Annotations (Mark Specific Points)
    Annotations are text labels that highlight a point on the chart — perfect for showing peaks, lows, or trends.

```python
import matplotlib.pyplot as plt

months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun']
sales = [200, 250, 300, 350, 400, 450]

plt.figure(figsize=(8,5))
plt.plot(months, sales, marker='o', color='teal', label='Sales')

# Highlight the highest point
max_sales = max(sales)
max_month = months[sales.index(max_sales)]
plt.annotate(
    f'Highest: {max_sales}',
    xy=(max_month, max_sales),
    xytext=('Apr', 420),
    arrowprops=dict(facecolor='black', shrink=0.05),
    fontsize=10,
    color='black'
)

plt.title('Sales Over Time with Annotation')
plt.xlabel('Month')
plt.ylabel('Sales ($)')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```

## ✅ The arrow and text highlight the best sales month.
    You can show multiple charts side-by-side using plt.subplot() or plt.subplots().

```python
import numpy as np

x = np.arange(1, 6)
sales = [200, 250, 300, 350, 400]
profit = [50, 70, 90, 120, 150]

plt.figure(figsize=(10,4))

# Chart 1: Line plot
plt.subplot(1, 2, 1)
plt.plot(x, sales, color='blue', marker='o')
plt.title('Sales')

# Chart 2: Bar chart
plt.subplot(1, 2, 2)
plt.bar(x, profit, color='green', edgecolor='black')
plt.title('Profit')

plt.tight_layout()  # Prevent overlap
plt.show()
```

### ✅ Output: two charts side-by-side — Sales and Profit.

## Subplots with plt.subplots() (Cleaner Way)
    This is the modern, preferred way to create multiple charts.

```python
import matplotlib.pyplot as plt

# ✅ Step 1: Create data
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun'] # → your x-axis labels
sales = [200, 250, 300, 350, 400, 450]              # → your y-axis values

# ✅ Step 2: Create subplots
fig, axes = plt.subplots(1, 2, figsize=(10,4))

# Left plot - Bar chart
axes[0].bar(months, sales, color='skyblue')
axes[0].set_title('Monthly Sales')
axes[0].set_ylabel('Sales ($)')
                                    # → refer to the two subplots created by plt.subplots(1, 2)
# Right plot - Line chart
axes[1].plot(months, sales, marker='o', color='orange')
axes[1].set_title('Sales Trend')

plt.suptitle('Sales Overview')      # → adds one big title for the whole figure
plt.tight_layout()                  # → prevents overlap
plt.show()
```

### ✅ axes[0] and axes[1] are your individual mini-plots.

## Customizing Fonts, Grid, and Axis

```python
plt.figure(figsize=(8,5))
plt.plot(months, sales, color='purple', marker='o')

plt.title('Customized Chart', fontsize=18, fontweight='bold', color='darkblue')
plt.xlabel('Month', fontsize=14)
plt.ylabel('Sales ($)', fontsize=14)

plt.grid(True, linestyle='--', color='gray', alpha=0.7)
plt.xticks(fontsize=12, rotation=45)
plt.yticks(fontsize=12)
plt.show()
```

### ✅ Fonts are larger, colors are clear, and the x-axis is rotated for better readability.

## Add Text and Shapes
    You can directly draw text or shapes like lines and boxes:

```python
plt.figure(figsize=(7,4))
plt.bar(months, sales, color='lightgreen', edgecolor='black')

plt.text(2, 380, 'Sales rising steadily!', fontsize=12, color='darkred')
plt.axhline(300, color='red', linestyle='--', linewidth=1.5, label='Target Line')

plt.legend()
plt.title('Annotated Bar Chart')
plt.show()
```

### ✅ Adds a text comment and a horizontal reference line.

## 📊 Professional Example: Dashboard Look

```python
plt.style.use('seaborn-v0_8')

fig, axs = plt.subplots(2, 2, figsize=(10,8))

axs[0, 0].plot(months, sales, color='blue', marker='o')
axs[0, 0].set_title('Sales Trend')

axs[0, 1].bar(months, sales, color='orange', edgecolor='black')
axs[0, 1].set_title('Monthly Sales')

axs[1, 0].hist(sales, bins=5, color='purple', edgecolor='black')
axs[1, 0].set_title('Sales Distribution')

axs[1, 1].pie(sales, labels=months, autopct='%1.1f%%', startangle=140)
axs[1, 1].set_title('Sales Share')

plt.suptitle('Company Performance Dashboard', fontsize=16, fontweight='bold')
plt.tight_layout()
plt.show()
```

### ✅ Creates a full dashboard layout with 4 different visualizations in one figure!
