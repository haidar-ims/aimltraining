```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.random.rand(50)
y = np.random.rand(50)
temp = np.random.randint(20,100,size=50)
plt.scatter(x,y,c=temp,cmap='plasma',s=120,edgecolor='black')  # cmap choices viridis, inferno, cool, hot, rainbow, grayscale
plt.colorbar(label='Temperature')
plt.title('Scatter Plot Colored by Temperature')
plt.xlabel('X-Random')
plt.ylabel('Y-Random')
plt.show()
```

```python
height_cm=[150, 155, 160, 165, 170, 175, 180, 185]
weight_kg=[50,53,56,59,62,66,68,72]
age_years=[20,22,25,28,30,35,38,40]

plt.figure(figsize=(8,5))

#<Figure size 800x500 with 0 Axes>
#<Figure size 800x500 with 0 Axes>

scatter = plt.scatter(height_cm, weight_kg, c=age_years, cmap='plasma', s=120, edgecolor='black')

plt.colorbar(scatter,label='Age (years)')

plt.title('Height vs Weight')
plt.xlabel('Height')
plt.ylabel('Weight')
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```

## Pie Plot Example
    plt.pie(x,labels='', colors='', autoopct='', startangle='', shadow='', explode='', radius='')

```python
depts=['HR','Web-Development','App-Development','Agile-Scrum','Testing']
expenses=[12000,25000,18000,15000,20000]

plt.figure(figsize=(7,7))

#<Figure size 700x700 with 0 Axes>
#<Figure size 700x700 with 0 Axes>

plt.pie(expenses, labels=depts,autopct='%1.1f%%', startangle=140,explode=(0,0.05,0,0,0),shadow=True)

plt.title('Department-wise Expense Report')
plt.show()
```

```python
plt.pie(
    expenses,
    labels=depts,                        # label for each slice
    colors=plt.cm.rainbow([0.1, 0.3, 0.5, 0.7, 0.9]),  # color map
    autopct='%1.1f%%',                   # show percentage with 1 decimal
    startangle=140,                      # rotate starting angle
    explode=(0, 0.05, 0, 0, 0),          # “pop out” 2nd slice
    shadow=True                          # add shadow for depth
)

plt.title('Department-wise Expense Report')
plt.show()
```

```python
sales_data=['Mobile','Clothing','Groceries','Home Decor']
expenses=[45000,30000,35000,20000]

plt.figure(figsize=(4,4))

plt.pie(expenses, labels=sales_data,
        colors = plt.cm.cool([0.1, 0.3, 0.6, 0.8]),
        autopct='%1.1f%%',
        startangle=140,
        explode=(0.05,0,0,0),
        shadow=True,
        textprops={'color':'black', 'fontsize':8, 'fontweight':'bold'})
# Able to specify with colors = ['#FF9999','#66B3FF','#99FF99','#FFCC99']
plt.title('Sales Report')
plt.savefig('categorieswisesale.png', dpi=300, bbox_inches='tight')
plt.show()
```

# Histogram Example

```python
marks=[45,60,80,55,62,90,89,50,56,60,80,55,62,98,89,58,94,56,67,83,78,73]
plt.style.use('ggplot')
plt.figure(figsize=(8,5))
plt.hist(marks, bins=7, color='skyblue', edgecolor='darkblue', alpha=0.9)
plt.title('Histogram for Student Marks')
plt.xlabel('Marks Range')
plt.ylabel('Number of Students')
plt.grid(color='blue', linestyle='--', linewidth=0.2)
plt.show()
```
