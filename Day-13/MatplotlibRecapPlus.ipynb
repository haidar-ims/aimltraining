```python
import matplotlib.pyplot as plt
```

```python
months = ['Jan','Feb','Mar','Apr']
sales = [2000,3000,2500,3600]
```

### plt.plot(x,y,color='color name', linestyle='line_style', linewidth=value, marker='marker symbol', label='label name')

```python
plt.plot(months,sales,color='skyblue', linestyle='--', linewidth=2, marker='^', label='Monthly Sales')
plt.xlabel('Months')
plt.ylabel('Sales per Month')
plt.title('Monthly Sales Report')
plt.xlim(0,6)
plt.ylim(2000,5000)
plt.grid(True)
plt.legend(loc='best', fontsize=10)
plt.show
```

```python
import matplotlib.pyplot as plt

categories = ['Electronics', 'Mobiles', 'Books', 'Groceries', 'Home Decoration']
sales = [50000, 60000, 30000, 90000, 40000]

bar_colors = ['red', 'blue', 'green', 'purple', 'orange']

plt.figure(figsize=(8,5))

plt.bar(categories, sales, color=bar_colors, edgecolor='black', label='Sales')

plt.xlabel('Categories')
plt.ylabel('Sales per Month')
plt.title('Category-wise Sales Report')

# Add horizontal gridlines only
plt.grid(axis='y', linestyle='--', alpha=0.5)

# Fix x-axis and y-axis ticks
plt.xticks(rotation=30, fontsize=12, color='magenta')

# Make y-ticks consistent and labeled correctly
plt.yticks([20000, 40000, 60000, 80000, 100000],
           ['20k', '40k', '60k', '80k', '100k'],
           fontsize=12, color='green')

plt.legend(loc='best', fontsize=10)

plt.tight_layout()
plt.show()
```

🧩 Explanation of fixes:

plt.show → plt.show()
You need parentheses to actually run the command.

plt.legend() needed a label
You added a legend but no labeled data. Fixed by adding label='Sales' in plt.bar(...).

plt.yticks(...) mismatch fixed
You gave 6 positions but only 4 labels.
Now it's aligned (5 positions, 5 labels).

```python
height_cm=[150, 155, 160, 165, 170, 175, 180, 185]
weight_kg=[50,53,56,59,62,66,68,72]

plt.figure(figsize=(7,5))
plt.scatter(height_cm, weight_kg, color='purple', s=100, alpha=0.7, label='Data Points')

plt.title('Height vs Weight')
plt.xlabel('Height (cm)')
plt.ylabel('Weight (kg)')

plt.xlim(140,190)
plt.ylim(40,90)

# Fix x-axis and y-axis ticks
plt.xticks(rotation=30, fontsize=12, color='magenta')

# Make y-ticks consistent and labeled correctly
plt.yticks([50, 55, 60, 65, 70, 75],
           ['50', '55', '60', '65', '70', '75'],
           fontsize=12, color='green')

plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```
