```python
import pandas as pd
import matplotlib.pyplot as plt
df=pd.read_csv('netflix_data.csv')
df=df.dropna(subset=['release_year'])
df['release_year']=df['release_year'].astype(int)
plt.figure(figsize=(8,5))
plt.hist(df['release_year'], bins=100,color='green', edgecolor='yellow', alpha=0.8)
plt.title('Netflix realeased Titles by Year')
plt.xlabel('Release Year')
plt.ylabel('Number of Titles')
plt.grid(color='blue', linestyle='--', linewidth=0.7)
plt.tight_layout()
plt.savefig('netflix_bar.pdf',dpi=300, bbox_inches='tight')
plt.show()
```

```python
df=df.dropna(subset=['release_year','type'])
```

```python
df['release_year']=df['release_year'].astype(int)
```

```python
movies=df[df['type']=='Movie']
shows=df[df['type']=='TV Show']
```

```python
plt.figure(figsize=(10,5))

plt.hist(
    [movies['release_year'], shows['release_year']],
    bins=30,
    color=['green','red'],
         edgecolor='yellow',
         alpha=0.8,
        label=['Movies','TV Shows']
)
plt.title('Netflix realeased Titles by Year')
plt.xlabel('Release Year')
plt.ylabel('Number of Titles')
plt.grid(color='blue', linestyle='--', linewidth=0.7)
plt.tight_layout()
plt.savefig('netflix_bar.pdf',dpi=300, bbox_inches='tight')
plt.show()
```

## Use netflix_data.csy dataset
### print how many Movies and TV Shows are available on Netflix
### Display Movies and TV Shows proportion in a pie plot
### Save file as pdf

```python
df = pd.read_csv('netflix_data.csv')
df=df.dropna(subset=['type'])
type_counts = df['type'].value_counts()
print(type_counts)
```

```python
plt.figure(figsize=(5,5))

plt.pie(
    type_counts,
    labels=type_counts.index,                        
    colors=plt.cm.rainbow([0.1, 0.3]),  
    autopct='%1.2f%%',                   
    startangle=140,                      
    explode=(0.05, 0),          
    shadow=True                         
)

plt.title('Netflix Movies and TV Shows Proportion')
plt.savefig('netflix_pieplot.pdf',dpi=300, bbox_inches='tight')
plt.show()
```
