```python
from sklearn.preprocessing import LabelEncoder

colors = ['Apple', 'Banana', 'Mango', 'Orange', 'Apple', 'Orange']
# ML algorithms cannot work directly with text, so we need to convert them to numbers.

# Initialize LabelEncoder
le = LabelEncoder() 
# LabelEncoder is a tool from scikit-learn.
# Its job is to map each unique category to a number.

# Fit and transform
encoded = le.fit_transform(colors)
# fit → identifies all unique categories (Apple, Banana, Mango, Orange).
# transform → converts each category to a number.

print(encoded)
```

```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder

data = pd.DataFrame ({'Region': ['North', 'South', 'East', 'West', 'South']})
ohe = OneHotEncoder (sparse_output=False)
encoded = ohe.fit_transform (data[['Region']])
encoded_df = pd.DataFrame(encoded, columns=ohe.get_feature_names_out(['Region']))
print(encoded_df)
```

## Linear Regression: Finds the best fitting straight line
##### Step 1: Collect the Data
##### Step2: Create the Model for LinearRegression
##### Step 3: Train the Model fit(X,y)
##### Step 4: Predict the Result predict([[value to_predict]])
##### Step 5: (optional): Visualize using plot of best fitted

```python
# Step 1: Collect the Data
# Example: hours studied vs marks obtained
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# Example data
hours_studied = np.array([1, 2, 3, 4, 5, 6, 7, 8]).reshape(-1, 1)  # study hours (features)
marks_obtained = np.array([20, 25, 30, 35, 50, 60, 70, 80])                # marks (target)

# Step 2: Create the Model
model = LinearRegression()

# Step 3: Train the Model
model.fit(X=hours_studied, y=marks_obtained)
```

```python
# Step 4: Predict
predicted_marks=model.predict([[9]])  # predict marks for 8 study hours
print("Predicted marks for 9 hours:", predicted_marks)
```

```python
hour_study=float(input('Number of hours study to predict marks'))
p_marks=model.predict([[hour_study]])
print(f'predicted Marks for {hour_study} hours is likely {p_marks}')
```
