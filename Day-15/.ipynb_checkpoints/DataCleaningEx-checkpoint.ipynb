```python
import pandas as pd
```

```python
data={'Students':['Amit','Anam','Chi','Sam','Ravi','Jai'],
      'Marks':['85',None,'65','75',None,'55']}
```

```python
df=pd.DataFrame(data)
print('Original Data')
df
```

#### Finds missing data

```python
print('Missing Data')
print('Number of Missing Values are:',df.isnull().sum())
```

```python
average_marks=df['Marks'].mean()
print(average_marks)
```

```python
df.fillna(average_marks, inplace=True)
```

```python
print('after fillna() call')
df
```

### Dropna() Example

```python
students_data={'Name': ['Amit', 'Anam', 'Chi', 'Sam', 'Ravi', 'Jai'], 'Marks': [85, None, 65, 75, None, 55]}
df=pd.DataFrame (students_data)
print('Original Data')
df
```

```python
df=df.dropna()
print('Dataset after dropna()')
df
```

## Example

```python
emp_data={
    "Name": ['Sam', 'Ali', 'Chang', 'Xi', 'Viz', 'Suhail'],
    "Age": [35, None, 25, 20, 24, None],
    "Salary": [50000, 30000, 20000, 40000, None, None],
}
df=pd.DataFrame(emp_data)
print('Original Data')
df
```

```python
print(df.isnull().sum())
```

```python
df_drop=df.dropna()
df_drop
```

```python
df_drop_salary=df.dropna(subset=['Salary'])
df_drop_salary
```

#### Download students_performance_dirty.csv.

#### Check the number of missing values in the dataset.

#### Drop rows where gender is missing.

#### Fill missing values in study_hours with the average study hours.

```python
df=pd.read_csv('students_performance_dirty.csv')
df
```

```python
df.dropna(subset=['gender'], inplace=True)
```

```python
df['study_hours'] = df['study_hours'].fillna(df['study_hours'].mean())
```

```python
print('Missing Data after cleaning:')
print(df.isnull().sum())
```
